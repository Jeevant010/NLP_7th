Ex 1. Represent following sentence in one-hot encoding .

```python
import numpy as np
sentence="IIIT Surat is a leading institute for computer science education"
print("original sentence: ",sentence)
print()

lower_sentence = sentence.lower()
print("LowerCase : ", lower_sentence)
print()

tokens = lower_sentence.replace(".", "").split()
print("Tokens : ", tokens)
print()

unique_words = list(set(tokens))
print("Unique Token : ", unique_words)
print()

sorted_words = sorted(unique_words)
print("Sorted Words : " , sorted_words)
vocab_size = len(sorted_words)
print("Vocab Size : ", vocab_size)
print()

word_indexings = { word : idx for idx, word in enumerate(sorted_words)}
print("Index Mappings : , word_indexings")
print()

one_hot_vectors = []
for word in sorted_words:
	vector = np.zeros(vocab_size, dtype=int)
	vector[word_indexings[word]] = 1
	one_hot_vectors.append(vector)

print("One Hot Encoders : ")
for word in sorted_words:
	print(f" {word:15s} -> {one_hot_vectors[word_indexings[word]]} ")
print()

one_hot_matrix = np.array(one_hot_vectors)
print(one_hot_matrix)
print()
print("Matrix shape : ", one_hot_matrix.shape)



```

Ex 2 Create a Bag of Words Model with Sklearn

```python
from sklearn.feature_extraction.text import CountVectorizer

sentence_1 = "IIIT Surat offers computer science education"
sentence_2 = "IIIT Surat offers computer engineering education"

documents = [sentence_1, sentence_2]
print("Document : ")

for i, doc in enumerate(documents, 1):
	print(f" Sentence {i} : {doc} ")
print()

vectorizer = CountVectorizer()

bow_matrix = vectorizer.fit_transform(documents)

print("Vocabulary : ")
for word, idx in sorted(vectorizer.vocabulary_.items(), key=lambda x: x[1]) :
	print(f" {idx:2d} : {word}")
print()

print(" Bag of Words ( Shape : ", bow_matrix.shape, " )")
print(bow_matrix.toarray())
print()

print("Interpretation")

features_names = vectorizer.get_feature_names_out()
for i, docs in enumerate(documents, 1):
	counts = bow_matrix.toarray()[i-1]
	print(f" Sentence {i} : {doc} ")
	for word, count in zip(features_names, counts):
		if count > 0:
			print(f" ' {word}' : {count} ")
```
